# はじめに

点から直線までの誤差には，$y$方向に測る**縦残差**と，直線に垂直な方向に測る**直交距離**があります．同じ点と同じ直線については，二つの量を簡単な式で換算できます．

このノートブックでは，点と直線の距離の公式を使い，
$$
\text{直交距離平方和}
=
\frac{\text{縦残差平方和}}{1+m^2}
$$
となる理由を確認します．ここで$m$は直線の傾きです．

## 到達目標

1. 縦残差と直交距離の違いを図形的に説明する．
2. 点と直線の距離の公式から，二つの誤差の関係を導く．
3. SymPyで解析的に値を求め，NumPyで全データ点の誤差を確認する．
4. QR回帰とPCAで最小化する誤差が異なる理由を説明する．

# 1　点と直線の二つの距離

点$P_i=(x_i,y_i)$と直線
$$
y=mx+b
$$
を考えます．同じ$x_i$における直線上の点は$Q_i=(x_i,mx_i+b)$なので，縦残差は
$$
r_i=y_i-(mx_i+b)
$$
です．縦残差の絶対値$|r_i|$は$y$方向に測った距離であり，一般には点から直線までの最短距離ではありません．

## 1-1　直線の一般形

直線の式を移項すると，
$$
mx-y+b=0
$$
となります．一般に，点$(x_i,y_i)$と直線$ax+cy+d=0$の距離は
$$
d_i=\frac{|ax_i+cy_i+d|}{\sqrt{a^2+c^2}}
$$
です．$a=m$，$c=-1$，$d=b$を代入すると，
$$
d_i
=\frac{|mx_i-y_i+b|}{\sqrt{m^2+1}}
=\frac{|y_i-(mx_i+b)|}{\sqrt{1+m^2}}
=\frac{|r_i|}{\sqrt{1+m^2}}
$$
となります．

## 1-2　平方和の関係

各点について二乗すると，
$$
d_i^2=\frac{r_i^2}{1+m^2}
$$
です．傾き$m$は全データ点で共通なので，総和を取ると，
$$
\sum_{i=1}^n d_i^2
=\frac{1}{1+m^2}\sum_{i=1}^n r_i^2
=\frac{\sum_{i=1}^n\{y_i-(mx_i+b)\}^2}{1+m^2}
$$
となります．したがって，固定した1本の直線について，直交距離平方和は縦残差平方和を$1+m^2$で割れば求められます．

# 2　1点での計算例

点$P=(3,7)$と，QR回帰で得られた直線
$$
y=\frac54x
$$
を考えます．縦残差は
$$
r=7-\frac54\cdot3=\frac{13}{4}
$$
です．直交距離は
$$
d=\frac{|13/4|}{\sqrt{1+(5/4)^2}}
=\frac{13}{\sqrt{41}}
$$
となります．SymPyで分数と平方根を保ったまま確認します．

````python
import sympy as sp
from IPython.display import Markdown, display

m = sp.Rational(5, 4)
b = sp.Integer(0)
x_i = sp.Integer(3)
y_i = sp.Integer(7)

vertical_residual = y_i - (m*x_i + b)
orthogonal_distance = (
    sp.Abs(vertical_residual) / sp.sqrt(1 + m**2)
)

display(Markdown(r'$r=$'))
display(vertical_residual)
display(Markdown(r'$d=$'))
display(sp.simplify(orthogonal_distance))
display(Markdown(r'$r^2/(1+m^2)=$'))
display(sp.simplify(vertical_residual**2 / (1 + m**2)))
````

# 3　全データ点での確認

次の6点と直線$y=\frac54x$について，縦残差平方和と直交距離平方和をNumPyで計算します．数値を4桁程度で表示し，
$$
\text{直交距離平方和}
=\frac{\text{縦残差平方和}}{1+m^2}
$$
となることを目視で確認してください．

## 3-1　NumPy配列を関数に渡すときの考え方

NumPy配列を`def`で定義した関数へ渡したとき，`def`そのものが自動的にPythonのloopを回すわけではありません．関数内の演算をNumPyの**配列演算**で記述しているため，ベクトル全体をまとめて処理できます．

例えば，

```python
import numpy as np

def sum_of_squares(v):
    return np.sum(v**2)
```

に`v = np.array([1, 2, 3])`を渡すと，`v**2`は各要素に二乗を適用して`[1, 4, 9]`を作り，`np.sum`はその全要素を足して`14`という1個の値にまとめます．このように複数の値を1個にまとめる演算を**reduction**と呼びます．

このノートブックの式は，次の順に処理されます．

| NumPyの式 | 処理 |
|:--|:--|
| `m*x` | スカラー`m`を`x`の各要素に掛ける |
| `y - (m*x + b)` | 同じ位置の要素どうしで縦残差を求める |
| `vertical_residuals**2` | 各縦残差を二乗する |
| `np.sum(vertical_residuals**2)` | 二乗した値をすべて足して1個の値にまとめる |

スカラー`m`や`b`を配列の各要素へ適用する仕組みを**broadcasting**と呼びます．この書き方ではPythonの`for`文は現れませんが，loopが消えたわけではありません．NumPyの内部で，主にコンパイル済みの高速な処理としてloopが実行されます．配列全体を対象としてこのように記述する方法を**vectorization**と呼びます．

NumPy配列の和には，Python組み込みの`sum(v)`より，`np.sum(v)`または`v.sum()`を使うのが自然です．2次元配列では，`axis`によって集約する方向を指定できます．

```python
A.sum()         # 全要素の和
A.sum(axis=0)   # 列ごとの和
A.sum(axis=1)   # 行ごとの和
```

なお，関数内にNumPyの配列演算を書けば何でも自動的にベクトル対応するわけではありません．例えば，配列`v`に対して通常の`if v > 0:`を書くと，複数の真偽値のどれを使うか決められないためエラーになります．配列を扱う関数では，使用する演算が配列に対応しているかを確認する必要があります．

````python
import numpy as np

x = np.array([3, -4, 7, 1, -4, -3], dtype=float)
y = np.array([7, -6, 8, -1, -1, -7], dtype=float)
m = 5/4
b = 0.0

vertical_residuals = y - (m*x + b)
orthogonal_distances = (
    np.abs(vertical_residuals) / np.sqrt(1 + m**2)
)
vertical_sse = np.sum(vertical_residuals**2)
orthogonal_sse = np.sum(orthogonal_distances**2)

print('縦残差 =', vertical_residuals.round(4))
print('直交距離 =', orthogonal_distances.round(4))
print('縦残差平方和 =', round(vertical_sse, 4))
print('直交距離平方和 =', round(orthogonal_sse, 4))
print('縦残差平方和/(1+m**2) =',
      round(vertical_sse/(1 + m**2), 4))
````

# 4　縦残差と直交距離のプロット

左図では各点から直線までの縦残差，右図では最短距離となる直交線分を描きます．直交射影点$(x_i^{\perp},y_i^{\perp})$は，
$$
x_i^{\perp}=\frac{x_i+m(y_i-b)}{1+m^2}，\qquad
y_i^{\perp}=mx_i^{\perp}+b
$$
で求められます．

````python
import matplotlib.pyplot as plt

x_perp = (x + m*(y - b)) / (1 + m**2)
y_perp = m*x_perp + b
x_line = np.linspace(-7.5, 7.5, 300)

fig, axes = plt.subplots(1, 2, figsize=(12, 5.5),
                         sharex=True, sharey=True)

for ax in axes:
    ax.scatter(x, y, marker='x', s=75, linewidths=2,
               color='tab:blue', label='data')
    ax.plot(x_line, m*x_line+b, color='firebrick',
            lw=2.2, label=r'$y=mx+b$')
    ax.axhline(0, color='0.75', lw=1)
    ax.axvline(0, color='0.75', lw=1)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_aspect('equal', adjustable='box')
    ax.grid(alpha=.2)

axes[0].scatter(x, m*x+b, s=35, color='tab:red',
                label='vertical projection')
for xi, yi in zip(x, y):
    axes[0].plot([xi, xi], [yi, m*xi+b],
                 color='0.35', lw=1.4)
axes[0].set_title('Vertical residuals')
axes[0].legend(fontsize=8)

axes[1].scatter(x_perp, y_perp, s=35, color='0.35',
                label='orthogonal projection')
for xi, yi, xpi, ypi in zip(x, y, x_perp, y_perp):
    axes[1].plot([xi, xpi], [yi, ypi],
                 color='0.35', lw=1.4)
axes[1].set_title('Orthogonal distances')
axes[1].legend(fontsize=8)

fig.suptitle('Vertical residuals and orthogonal distances')
fig.tight_layout()
plt.show()
````

# 5　QR回帰とPCAが異なる理由

固定した傾き$m$については，縦残差平方和を$1+m^2$で割れば直交距離平方和へ換算できます．しかし，最適な傾き$m$を探すときには，分母$1+m^2$も$m$とともに変化します．したがって，
$$
\mathop{\mathrm{minimize}}_m\sum_i\{y_i-(mx_i+b)\}^2
$$
と
$$
\mathop{\mathrm{minimize}}_m\frac{\sum_i\{y_i-(mx_i+b)\}^2}{1+m^2}
$$
は同じ最適化問題ではありません．

QR回帰は$y$方向の縦残差平方和を最小化します．一方，PCAは点から直線までの直交距離平方和を最小化します．この違いにより，両者が与える直線の傾きは一般には一致しません．

## まとめ

直線$y=mx+b$に対して，
$$
r_i=y_i-(mx_i+b)，\qquad
d_i=\frac{|r_i|}{\sqrt{1+m^2}}
$$
であり，
$$
\sum_i d_i^2=\frac{\sum_i r_i^2}{1+m^2}
$$
です．この関係は，同じ直線に対する二つの誤差を換算する式です．直線そのものを最適化する場合には，QR回帰とPCAで目的関数が異なることに注意してください．
