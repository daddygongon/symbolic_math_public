# 微積分
## ソフトマックス関数の概形(15点)
ソフトマックス関数
    \begin{equation*}
      f(x) = \frac{1}{1+e^{-x}}
    \end{equation*}
    の増減，極値，凹凸を調べ，曲線$y=f(x)$の概形を描け．
   

```maple
with(plots):
```

```maple
f:=unapply(1/(1+exp(-x)),x);
```

```maple
p1:=plot(f(x),x):
p2:=plot(diff(f(x),x),x):
p3:=plot(diff(f(x),x,x),x):
with(plots):
display(p1,p2,p3);
```

```maple
solve(diff(f(x),x,x)=0,x);
```

| x      | $-\infty$ | $\cdots$   |      0 | $\cdots$   | $\infty$ |
| :----: |    :----  | :----     | :---- | :----     |   :----|
| f(x)   |         0 | $\nearrow$ |    0.5 | $\nearrow$ |        0 |
| f'(x)  |         0 | +          |      + | +          |        0 |
| f''(x) |         0 | +          |      0 | -          |        0 |
