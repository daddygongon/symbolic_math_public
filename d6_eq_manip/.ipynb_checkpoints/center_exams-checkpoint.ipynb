# 課題１　数値改変：代入，展開，方程式
## 数値改変課題

`eq_manip.ipynb`の演習2では，$P=x(x+3)(2x-3)$について，$x=a+1$の場合を扱った．ここでは$x=a+2$として問題を解け．
ただし，数値改変に伴い，解答欄は意味をなさない可能性がある．

## 代入，展開，方程式（元問題）

$P=x(x+3)(2x-3)$とする．また，$a$を定数とする．$x=a+1$のときの$P$の値は
$$
2a^3+\fbox{ア}a^2+\fbox{イ}a-\fbox{ウ}
$$
である．

$x=a+1$のときの$P$の値と，$x=a$のときの$P$の値が等しいとする．このとき，$a$は
$$
3a^2+\fbox{エ}a-\fbox{オ}=0
$$
を満たす．したがって
$$
a=\frac{\fbox{カキ}\pm\sqrt{\fbox{クケ}}}{\fbox{コ}}
$$
である．（2008年度大学入試センター試験数学Iより抜粋）


## ヒント

代入して方程式の解を求める典型的な問題です．一歩ずつ，出力しながら変形するのがコツです．

```
代入:  subs
展開:  expand
方程式の解: solve(eq, a)
```

SymPyでは，`solve(eq, a)`によって$eq=0$を満たす$a$を求めます．したがって，`eq`には左辺と右辺の差を渡します．

# 課題２　二次関数の頂点，零点，接線

$C_1$を次の二次関数のグラフとする．
$$
C_1:f(x)=x^2+3x-2
$$

## (a)
$C_1$の頂点を求める．$C_1$の微分をとり，その傾きが0となる点は
   $$
   x_0=\fbox{ア}
   $$
   と求められる．この頂点の座標は
   $$
   (\fbox{イ},\fbox{ウ})
   $$
   である．

## (b)
$C_1$が$x$軸と交わる座標を求める．
$$
f(x)=0
$$を$x$について解くことで，
$$
x=\fbox{エ},\fbox{オ}
$$
が求められる．



## (c)

$C_1$に任意の点$x=x_1$で接する直線の方程式を求める．$C_1$の傾きを表す関数$m(x)$は
$f(x)$の微分により，
$$
m(x)=\fbox{カ}
$$
と求められる．$x=x_1$ における折線$l$の方程式は，
   $$
   l:y=m(x_1)(x-x_1)+f(x_1)
   $$
   より，Python codeとしては，

   ```python
   def l_y(x, x1):
       m = diff(f,x)
       eq = m.subs({x:x1})*(x-x1)+f.subs({x:x1})
       return eq
   ```

   となる．$x_1=2$における接線が
   $$
   y=7x-6
   $$
   となることをcodeで導け．
   また$C_1$と接線を同時プロットは図の通りとなることをcodeで確認せよ．．


````python
from sympy import *
init_printing()
x = symbols('x')

f = x**2+3*x-2

%matplotlib inline
plot(f, 7*x-6, (x,-4,4))
````

# 課題3　二つの放物線の頂点

$a,b$を実数とし，二次関数
$$
y=4x^2-8x+5\qquad(1)
$$
$$
y=-2(x+a)^2+b\qquad(2)
$$
の表す放物線をそれぞれ$C_1,C_2$とする．$C_1$の頂点と$C_2$の頂点が一致するとき，
$$
a=\fbox{アイ},\qquad b=\fbox{ウ}
$$
である．空欄ア〜ウを求め，二つの関数を同時にプロットして確認せよ．（2001年度大学入試センター試験数学I・数学Aより抜粋）

````python
from sympy import *
#init_session()
init_printing()

line_1 = 4*x**2-8*x+5
line_2 = -2*(x-1)**2+1
plot(line_1, line_2, (x,-1,3))
````

## 別解
こういった問題では，判別式を使った解法を指導しているというTAさんからの指導(?)があったんで，やって見た．


````python
from sympy import *
# init_session()

````

````python
x, y,a,b = symbols('x y a b')
y1 = 4*x**2-8*x+5
y2 = -2*(x+a)**2+b
````

````python
y3 = expand(y1-y2)
y3
````

````python
cc=y3.coeff(x,0)
bb=y3.coeff(x,1)
aa=y3.coeff(x,2)
print(aa,bb,cc)
````

````python
discriminant = expand(bb**2-4*aa*cc)
print("判別式:", factor(discriminant))

# 共通頂点では2曲線が接するので，y1-y2=0の重解がy1の頂点と一致します．
x0_alt = solve(diff(y1,x),x)[0]
x_touch = simplify(-bb/(2*aa))
print("y1の頂点のx座標:", x0_alt)
print("y1-y2=0の重解:", x_touch)

solutions_alt = solve(
    [discriminant, x_touch-x0_alt],
    [a,b],
    dict=True
)
solutions_alt
````

> **注：`dict=True`について**
>
> `solve`で`dict=True`を指定すると，解は「変数と値の対応を表す辞書」のリストとして返されます．今回の結果`[{a: -1, b: 1}]`は，$a=-1$，$b=1$という解が1組あることを表します．複数の解がある場合は，辞書が複数並びます．
>
> 例えば，最初の解から$a,b$の値を取り出すには，次のように書きます．
>
> ```python
> solution = solutions_alt[0]
> solution[a]
> solution[b]
> ```
>
> 辞書は式への代入にもそのまま使えます．
>
> ```python
> y2.subs(solution)
> ```
>
> `dict=True`を付けない場合は，問題によって値，リスト，タプルなど返り値の形式が変わることがあります．連立方程式では，変数と解の対応が明確になる`dict=True`が扱いやすい指定です．
>
> 辞書のキーは文字列`"a"`，`"b"`ではなく，`a, b = symbols('a b')`で宣言したSymPyの`Symbol`です．このため，宣言済みの`a,b`を`solution[a]`，`solution[b]`のように直接キーとして使えます．同じ`Symbol`をキーとする辞書なので，`subs`にもそのまま渡せます．

# 課題4　数値改変：文字係数を含む二次関数

## 数値改変課題

`eq_manip.ipynb`の例題では，二次関数が点$(-2,6)$を通る場合を扱った．ここでは通過点を$(-3,8)$として問題を解け．
ただし，数値改変に伴い，解答欄は意味をなさない可能性がある．

## 文字係数を含む二次関数（元問題）

$a,b$を定数とし，$a\neq0$とする．二次関数
$$
y=ax^2-bx-a+b
$$
のグラフが点$(-2,6)$を通るとする．このとき，
$$
b=-a+\fbox{ア}
$$
であり，グラフの頂点の座標を$a$を用いて表すと
$$
\left(\frac{-a+\fbox{イ}}{\fbox{ウ}a},-\frac{(\fbox{エ}a-\fbox{オ})^2}{\fbox{カ}a}\right)
$$
である．（2008年度大学入試センター試験数学Iより抜粋）

## ヒント

頂点を代数式で表すには，式の整理が必要です．`simplify`だけで期待した形にならない場合は，`together`，`fraction`，`expand`，`factor`を組み合わせます．

```
代入:   subs
展開:   expand
簡単化: simplify
通分:   together
因数分解:factor
分数:   fraction
分母:   d(enominator)
分子:   n(umerator)
```
