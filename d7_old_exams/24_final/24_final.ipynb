# 問3 センター試験(20 点)

第1回試行調査：数学II・B 第1問[1]


$a$を定数とする．
座標平面上に，原点を中心とする半径5の円$C$と，
直線$l: x+y=a$がある．

$C$と$l$が異なる2点で交わるための条件は，
$$
-\fbox{ ア }\sqrt{\fbox{ イ }}<a<\fbox{ ア }\sqrt{\fbox{ イ }} \,\cdots\,(1)
$$
である．(1)の条件を満たすとき，
$C$と$l$の交点の一つをP$(s,t)$とする．
このとき，
$$
st = \frac{a^2-\fbox{ ウエ }}{\fbox{ オ }}
$$
である．
