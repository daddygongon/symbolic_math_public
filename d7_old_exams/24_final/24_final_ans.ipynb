# 問１　微積分

## ローレンツ関数(15 点) 
次のローレンツ関数
$$
f(x) = a_1 + \frac{a_2}{a_3+(x-a_4)^2}
$$
を
$a_1 = 0.8, a_2 = 15, a_3 = 1.6, a_4 = 4.0$の場合でプロットせよ．

```python
from sympy import *

x = symbols('x')
a_1 = 0.8
a_2 = 15
a_3 = 1.6
a_4 = 4.0

f= a_1 + a_2/(a_3+(x-a_4)**2)
```

```python
from sympy.plotting import plot
p1 = plot(f, (x,-20,20),
         legend=True, show=False)
p1[0].line_color = 'r'
p1.show()
```

## ローレンツ関数(15 点)

問1.1の関数$f(x)$の1次微分，2次微分を求めて，最大点，変曲点の$x$座標を求めよ．
また，それら関数を同時にプロットせよ．
さらに，元の関数$f(x)$の特徴を述べよ．

```python
df = diff(f,x)
df
```

```python
d2f = diff(f,x,x)
d2f
```

```python
plot(f, df, d2f)
```

```python
solve(df,x)
```

```python
s2=solve(d2f,x)
s2
```

```python
print( s2[0]-4.0, s2[1]-4.0)

# 最大値がx=a4の時で，　変曲点はその前後０．７３程度にあり，元の関数はx=a4を中心とした左右対称の関数である．
```

# 問２　線形代数

## 固有値(15 点)

次の行列$A$の固有値と固有ベクトルを求めよ．
$$
A=\left(
    \begin{array}{@{\,}c@{\,}c@{\,}c@{\,}}
      1 & -1 & -1 & -1 \\
      0 & 2 & -1 & -1 \\
      0 & 0 & 3 & -1 \\
      0 & 0 & 0 & 4 \\
\end{array}
    \right)
$$

```python
from sympy import *

#init_session()
#init_printing(use_unicode=True)
init_printing()
A = Matrix([[1,-1,-1,-1],[0,2,-1,-1],[0,0,3,-1],[0,0,0,4]])
A
```

```python
eigens=A.eigenvects()
eigens
```

```python
p_e=A.diagonalize()
p_e
```

```python
p_e[0].inv()*A*p_e[0]
```

# 問3 センター試験(20 点)

第1回試行調査：数学II・B 第1問[1]


$a$を定数とする．
座標平面上に，原点を中心とする半径5の円$C$と，
直線$l: x+y=a$がある．

$C$と$l$が異なる2点で交わるための条件は，
$$
-\fbox{ ア }\sqrt{\fbox{ イ }}<a<\fbox{ ア }\sqrt{\fbox{ イ }} \,\cdots\,(1)
$$
である．(1)の条件を満たすとき，
$C$と$l$の交点の一つをP$(s,t)$とする．
このとき，
$$
st = \frac{a^2-\fbox{ ウエ }}{\fbox{ オ }}
$$
である．

```python
from sympy import *

five = 5 # 5
a, b, c, s, t= symbols('a b c s t')

# x + 2*y - a = 0 <=> bx + cy -a =0 
a=a
b=1
c=1 # 1
solve(a/sqrt(b**2+c**2)-five,a)
```

```python
eq1=s**2+t**2 - five**2
eq2=b*s+c*t-a
```

```python
s1 = solve({eq1, eq2},{s,t})
s1
```

```python
simplify( expand( s.subs(s1[0]) * t.subs(s1[0]) ) )
```

```python
expand( s.subs(s1[1]) * t.subs(s1[1]) )
```
